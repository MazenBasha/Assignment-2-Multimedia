Load `outputs/metrics_mode_*.json` and render the comparison tables
and figures used in the report.

Run after the full pipeline (`bash scripts/run.sh`).

```python
import json, pandas as pd
a = json.load(open('../outputs/metrics_mode_a.json'))
b = json.load(open('../outputs/metrics_mode_b.json'))
pd.DataFrame({k: v for k, v in a.items() if k in ('rag','baseline')}).T
```

```python
pd.DataFrame({k: v for k, v in b.items() if k in ('rag','baseline')}).T
```

```python
import matplotlib.pyplot as plt
per = a.get('rag', {}).get('chexbert_per_label')
if per:
    labels = list(per.keys())
    rag_v  = [a['rag']['chexbert_per_label'][l]      for l in labels]
    base_v = [a['baseline']['chexbert_per_label'][l] for l in labels]
    x = range(len(labels))
    fig, ax = plt.subplots(figsize=(10,4))
    ax.bar([i-0.2 for i in x], rag_v,  width=0.4, label='RAG')
    ax.bar([i+0.2 for i in x], base_v, width=0.4, label='Baseline')
    ax.set_xticks(list(x)); ax.set_xticklabels(labels, rotation=45, ha='right')
    ax.set_ylabel('CheXbert F1'); ax.legend(); plt.tight_layout()
```
